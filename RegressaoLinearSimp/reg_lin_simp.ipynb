### Analise de regressão

É uma abordagem estatística que busca investigar e quantificar as relações entre variáveis.Ela é usasa para entender como variavél dependente (target)está relocianada a uma ou mais variáveis independentes (fatores que acreditamos influenciar a variável dependente)

Essa té cnica permite construir um modelo matématico geralmente na forma de equanção linear, que representa essa relação. Ao ajustar o modelo aos dados observados, podemos estimar os parâmetros e entender como as mudanças nas variáveis independentes afetam a variável dependente.

#### Tipos de regressão 

Linear Simples: envolve uma variável independente e uma dependente, sendo representada por uma linha reta

Linear  Multipla: Usada geralmente para prever uma variável dependente, resultando em um modelo linear mais complexo

Logítisca : Usada para prever probabilidades em problemas de classificação (Prever se um cliente fará compra sim/não)

Polinomial: permite modelar relações não lineares entre variáveis. A regressão polinomial usa uma curva (um polinômio) para se ajustar aos dados (Previsão de vendas com base no clima/temperatura)

Polisson

Ridge,Lasso e ElaticNet

Gamma

Beta
