##### Tipos de aprendizado

- Supervisionado:
   
   É usado um dataset já rotulado 
- Não supervisionado:
    
    Aprende com dados não rotulados, através da idetificações de padrões e outras estruturas
- Semi supervisionado:
    
    combina dados rotulados e não rotulados
- Por Reforço:
    
    Agentes aprendem a tomar ações com base em um ambiente para maximizar algum tipo de recompensa

Overfiting ocorre quando um modelo de aprendizado de máquina aprende a relação entre as variaveis, nos dados de treinamento com muito detalhe, incluindo ruído nos dados, podendo ser preciso nas previsões de treinamento, mas não sendo capaz de generalizar em dados novos

Underfit ocorre quando um modelo de aprendizado de máquina é muito simples para aprender a relação entre as variáveis nos dados de treinamento , isspo pode resultar em um modelo que não é capaz de fazer previsões para dados novos


   Viés : é o erro sistematico que um modelo comete ao aprender a partir de dados. Ele ocorre quando o modelo não é capaz de aprender a relação entre as variaveis

Variância é a variabilidade dos resultados de um modelo a o ser aplicado a diferentes conjuntos de dados. ele ocorre quando o modelo é muito complexo ou quando os dados de treinamento são insuficientes

Trad off entre viés e variância descreve a relação entre a capcidade de um modelo de aprender a partir de dados e sua capacidade generalizar para dados novos
