# Análise bivariada

### Análise de correlação
Analisar a correlação de duas variaveis

```python
# Hipotese: Clientes com Contrato do Tipo Mensal são mais propensos a cancelar o serviço
# Construir uma tabela de Contigencia entre tipo de contrato e Churn

pd.crosstab(df_churn.Churn, df_churn.Contract, margins= True, normalize= 'index', margins_name='Total')
```

```python
# Avaliar a correlação entre duas variáveis categoricas (qualitativas)

#executar um teste de hipotese chamado Qui-Quadrado (Chi-Square) DE Pearson
#num teste de hipotes, duas hipoteses são formuladas: 
# a hipotese nula (H0) : as duas variáveis são independentes
# hipotese alternativa (H1): as duas variáveis não são independentes


# O teste serve para confirmar ou rejeitar a hipotese nula
# Quando a probabilidade de observamos oH) é inferior a 0.05 (rejeitamos), rejeitamos a hipotese nula
# recusamos e seguimos com a complementar (H1)
```

```python
#Gerar um DF da Crosstab (sem totais)
df_crosstab_churn_contract = pd.crosstab(df_churn.Churn, df_churn.Contract)
```

```python
print(df_crosstab_churn_contract)
```

```python
# Calcular os Scores e P-Value

from scipy.stats import chi2_contingency
```

```python
chi_scores_churn_contract = chi2_contingency(df_crosstab_churn_contract)

print (chi_scores_churn_contract)
```

```python
scores_churn_contract = pd.Series(chi_scores_churn_contract[0])
```

```python
pvalues_churn_contract = pd.Series(chi_scores_churn_contract[1])
```

```python
# Apresentar numeros com decimais sem notação científica

pd.set_option('display.float_format', lambda x : '%.15f' % x)
```

```python
# Criar um DF com os resultados

df_chi_scores_churn_contract = pd.DataFrame({'Qui2': scores_churn_contract, 'P-Value': pvalues_churn_contract})
```

```python
# Analisar os scores e p-values
df_chi_scores_churn_contract
```

Conforme o teste estatisico (Chi-Square), o P-Value <= 0.05, desta forma rejeitamos a hiótes nula ou seja as variaveis não sçao independentes. Pelo Qui2 alto podemos afirmar que há uma forte correlação

```python
# Avalia a correlação entre uma variavel qualitativa e uma quantitativa
# Usar o mesmo tese de Chi-Square, mas transofrmar a variavel quantitativa em qualitativa
# Hipotese: Cliente com menos de 6 meses de contrato é mais propenso a cancelar o serviço

# Criar uma coluna nova no Df com base numa condição de outra coluna

# Coluna = TempoMenor6meses

import numpy as np
```

```python
# Criar coluna nova

df_churn['TempoMenor6meses'] = np.where(df_churn.tenure < 6, 'Sim', 'Não')
```

```python
df_churn.head()
```

```python
# gerar um DF da Crosstab (sem totais)
df_crosstab_churn_tenure = pd.crosstab(df_churn.Churn, df_churn.TempoMenor6meses)

df_crosstab_churn_tenure.head()
```

```python
# Calcular os Scores e P-Value

chi_scores_churn_tenure = chi2_contingency(df_crosstab_churn_tenure)

print(chi_scores_churn_tenure)
```

```python
scores_churn_tenure = pd.Series(chi_scores_churn_tenure[0])
pvalues_churn_tenure = pd.Series(chi_scores_churn_tenure[1])
```

```python
#criar um DF com os resultados

df_chi_scores_churn_tenure = pd.DataFrame({'Qui2': scores_churn_tenure, 'P-Value': pvalues_churn_tenure})
```

```python
# analisar os resultados
```

```python
df_chi_scores_churn_tenure
```

Como o  P-Value <= 0.05, rejeitamos a H0, ou seja as variaveis não são independentes. E vale mencionar que com base do Qui2 resultante das análises, a correlação entre Churn e Tempo de Contrato < 6 meses é menos que a correlação entre churn e tipo de contrato

```python
#Correlação entre duas variáveis Numéricas
# Tenure com TotalCharges
# A intuição é que clientes com mais tempo de contrato maior o valor pago

# Correlaçãpo entre duas variáveis numéricas - Tenure e TotalCharges - Pearson

df_churn.tenure.corr(df_churn.TotalCharges)
```

```python
# Correlaçãpo entre duas variáveis numéricas - Tenure e TotalCharges - Spearman

df_churn.tenure.corr(df_churn.TotalCharges, method='spearman')
```

Há uma forte correlação entre Tenure e Total Charhes usando metodos estatisticos de correlação
