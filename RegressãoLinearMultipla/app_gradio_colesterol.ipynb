```python
import gradio as gr 
import joblib
import pandas as pd
```

```python
modelo = joblib.load('./model_regre_colesterol.pkl')
```

```python
def predict(grupo_sanguineo, fumante, nivel_atividade_fisica,idade,peso,altura):
    _fumante = 'Sim' if fumante else 'Não'
    predicao_individual ={
    'grupo_sanguineo': grupo_sanguineo,
    'fumante': _fumante,
    'nivel_atividade_fisica': nivel_atividade_fisica,
    'idade': idade,
    'peso': peso,
    'altura': altura
    }
    sample_df = pd.DataFrame(predicao_individual, index=[1])
    colesterol = modelo.predict(sample_df)
    return colesterol.reshape(-1)[0]
    

```

```python
demo = gr.Interface(
    fn=predict, 
    inputs=[
        gr.Radio(["A", "B", "AB", "O"]),
        'checkbox',
        gr.Radio(['Baixo','Moderado', 'Alto']),
        gr.Slider(20, 80, step=1),
        gr.Slider(40, 150, step=1),
        gr.Slider(150, 200, step=1)
    ], outputs='number'
)
demo.launch()
```
